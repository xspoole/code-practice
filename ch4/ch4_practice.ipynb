### Problem 1

Say you have a list value like this: `spam = ['apples', 'bananas', 'tofu', 'cats']`. Write a function that takes a list value as an argument and returns a string with all the items separated by a comma and a space, with *and* inserted before the last item. For example, passing the previous `spam` list to the function would return `'apples', 'bananas', 'tofu', 'cats'`. But your function should be able to work with any list value passed to it. Be sure to test the case where an empty list [] is passed to your function.

```python
def list_to_str(list):
    try:
        if len(list) > len(list) - 1:
            for i in range(len(list) - 1): # access every value in the list except for the last one
                print(list[i], end = ', ') # if you print(str(i)), only the numbers will be returned, end = '' means no new line
        print('and' + ' ' + list[-1]) # this cannot be in an elif statement otherwise it will never be accessed! [-1] is the last value in the list
    
    except:
        print('Please enter a list with values!!')
```

```python
list = 'erm ackshully'

list_to_str(list)
```
